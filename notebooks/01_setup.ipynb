```python
# =====================================================
# TAREFA 8: Criar estrutura de armazenamento de dados
# (Unity Catalog Volume - substitui o DBFS /mnt/ tradicional)
# =====================================================

print("=" * 60)
print("Criando estrutura de dados (Unity Catalog Volume)")
print("=" * 60)

# 1. Criar um Volume dentro do catálogo/schema padrão do workspace
#    (catalog "workspace", schema "default" já existem por padrão)
spark.sql("""
CREATE VOLUME IF NOT EXISTS workspace.default.nyc_taxi
COMMENT 'Armazena os dados do projeto NYC Taxi Analytics'
""")

print("✅ Volume 'nyc_taxi' criado (ou já existente)")

# 2. Definir o caminho-base do volume
VOLUME_PATH = "/Volumes/workspace/default/nyc_taxi"

# 3. Criar a estrutura de pastas dentro do volume
pastas = [
    f"{VOLUME_PATH}/raw/csv",
    f"{VOLUME_PATH}/raw/delta",
    f"{VOLUME_PATH}/processed/cleaned",
    f"{VOLUME_PATH}/processed/featured",
    f"{VOLUME_PATH}/processed/combined",
    f"{VOLUME_PATH}/analytics/aggregations",
    f"{VOLUME_PATH}/ml/models",
    f"{VOLUME_PATH}/ml/predictions",
    f"{VOLUME_PATH}/ml/experiments",
]

for pasta in pastas:
    dbutils.fs.mkdirs(pasta)
    print(f"✅ Criada: {pasta}")

print("\n" + "=" * 60)
print("✅ Estrutura de dados criada com sucesso!")
print("=" * 60)

# 4. Validar - listar o que foi criado
print("\n📂 Conteúdo do volume raiz:")
display(dbutils.fs.ls(VOLUME_PATH))
```
